```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

import warnings
warnings.filterwarnings("ignore")

plt.rcParams["figure.figsize"] = (10,6)
sns.set_style("whitegrid")
```

```python
import pandas as pd

df2015 = pd.read_csv("2015.csv")
df2016 = pd.read_csv("2016.csv")
df2017 = pd.read_csv("2017.csv")
df2018 = pd.read_csv("2018.csv")
df2019 = pd.read_csv("2019.csv")

df2015["Year"] = 2015
df2016["Year"] = 2016
df2017["Year"] = 2017
df2018["Year"] = 2018
df2019["Year"] = 2019

df = pd.concat([df2015, df2016, df2017, df2018, df2019],ignore_index=True,sort=False)

print("Dataset Loaded Successfully")
print("Total Rows :", df.shape[0])
print("Total Columns :", df.shape[1])

df.head()
```

```python
print("First 5 Rows")
display(df.head())

print("\nLast 5 Rows")
display(df.tail())
```

```python
print("Shape of Dataset :", df.shape)

print("\nColumns")

for col in df.columns:
    print(col)

print("\nData Types")
print(df.dtypes)

print("\nDataset Information")
df.info()
```

```python
print("Statistical Summary")

display(df.describe())

print("\nStatistical Summary Including Object Columns")

display(df.describe(include='all'))
```

```python
print("Missing Values")

missing = df.isnull().sum()

display(missing)

print("\nTotal Missing Values :", missing.sum())
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

for col in numeric_columns:
    df[col].fillna(df[col].mean(), inplace=True)

object_columns = df.select_dtypes(include='object').columns

for col in object_columns:
    df[col].fillna(df[col].mode()[0], inplace=True)

print("Missing Values After Cleaning")

display(df.isnull().sum())
```

```python
duplicates = df.duplicated().sum()

print("Duplicate Rows :", duplicates)
```

```python
df.drop_duplicates(inplace=True)

print("Shape After Removing Duplicates")

print(df.shape)
```

```python
numeric_df = df.select_dtypes(include=np.number)

print("Mean")
display(numeric_df.mean())

print("\nMedian")
display(numeric_df.median())

print("\nMode")
display(numeric_df.mode().iloc[0])

print("\nStandard Deviation")
display(numeric_df.std())

print("\nVariance")
display(numeric_df.var())
```

```python
numeric_df = df.select_dtypes(include='number')

print("Correlation Matrix")

display(numeric_df.corr())
```

```python
import os

if not os.path.exists("Graphs"):
    os.makedirs("Graphs")

print("Graphs folder created successfully.")
```

```python
columns = ["Happiness Score","Economy (GDP per Capita)","Family","Health (Life Expectancy)","Freedom","Trust (Government Corruption)","Generosity"]

corr = df[columns].corr()

plt.figure(figsize=(10,7))

sns.heatmap(corr,annot=True,cmap="coolwarm",fmt=".2f",linewidths=0.5)
plt.savefig("Graphs/Heatmap.png", dpi=300, bbox_inches="tight")

plt.title("Correlation Heatmap")
plt.show()
```

```python
score_column = [col for col in df.columns if "Score" in col or "score" in col][0]

top10 = df.nlargest(10, score_column)

display(top10)
```

```python
bottom10 = df.nsmallest(10, score_column)

display(bottom10)
```
